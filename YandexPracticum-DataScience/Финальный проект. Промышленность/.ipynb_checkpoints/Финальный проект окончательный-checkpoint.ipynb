**Описание проекта: промышленность**

Чтобы оптимизировать производственные расходы, металлургический комбинат ООО «Так закаляем сталь» решил уменьшить потребление электроэнергии на этапе обработки стали. Вам предстоит построить модель, которая предскажет температуру стали.

Описание этапа обработки

Сталь обрабатывают в металлическом ковше вместимостью около 100 тонн. Чтобы ковш выдерживал высокие температуры, изнутри его облицовывают огнеупорным кирпичом. Расплавленную сталь заливают в ковш и подогревают до нужной температуры графитовыми электродами. Они установлены в крышке ковша. 

Из сплава выводится сера (десульфурация), добавлением примесей корректируется химический состав и отбираются пробы. Сталь легируют — изменяют её состав — подавая куски сплава из бункера для сыпучих материалов или проволоку через специальный трайб-аппарат (англ. tribe, «масса»).

Перед тем как первый раз ввести легирующие добавки, измеряют температуру стали и производят её химический анализ. Потом температуру на несколько минут повышают, добавляют легирующие материалы и продувают сплав инертным газом. Затем его перемешивают и снова проводят измерения. Такой цикл повторяется до достижения целевого химического состава и оптимальной температуры плавки.

Тогда расплавленная сталь отправляется на доводку металла или поступает в машину непрерывной разливки. Оттуда готовый продукт выходит в виде заготовок-слябов (англ. slab, «плита»).
Описание данных

Данные состоят из файлов, полученных из разных источников:

    data_arc_new.csv — данные об электродах;
    data_bulk_new.csv — данные о подаче сыпучих материалов (объём);
    data_bulk_time_new.csv — данные о подаче сыпучих материалов (время);
    data_gas_new.csv — данные о продувке сплава газом;
    data_temp_new.csv — результаты измерения температуры;
    data_wire_new.csv — данные о проволочных материалах (объём);
    data_wire_time_new.csv — данные о проволочных материалах (время).
    
Во всех файлах столбец key содержит номер партии. В файлах может быть несколько строк с одинаковым значением key: они соответствуют разным итерациям обработки.

**Активная мощность**

Эта мощность, потребляемая нагрузкой типа обычного сопротивления. Как правило, это все нагревательные приборы (бойлеры, обычные электроплитки, электро калориферы и т.п.). Потребляемая мощность этих приборов полностью активная. В этих приборах электрическая энергия безвозвратно и полностью преобразуется в другой вид энергии (тепловую и другие).

Активная мощность обозначается буквой P и измеряется в ваттах (Вт).

Величина активной мощности, потребляемой такими приборами считается просто — умножением напряжения в розетке на ток, протекающей в цепи включенного нагревательного прибора:

P = U * I

Тут всё просто. Нагрузка пассивна, постоянна, никаких неожиданностей.

Замечу, что в цепях постоянного тока существует только активная мощность, поскольку значение мгновенной и средней мощности там совпадают.

**Реактивная мощность**

Если включить в сеть переменного тока не нагревательный прибор, а, например, электромагнит, то помимо активной, в цепи возникает реактивная энергия, которая с частотой переменного тока то потребляется прибором, то возвращается обратно в сеть. Эта энергия переносится от источника к электромагниту и обратно дважды за период, каждую четверть периода меняя направление.

Это происходит из-за того, что при потреблении электроэнергии, например, обмоткой магнита, каждый полупериод в нём происходит временное запасание энергии в магнитном поле катушки, и последующая отдача её назад, из-за чего происходит рассинхронизация синусоид величин напряжения и тока в сети.

Изменения тока в цепи отстаёт от соответствующих синусоидальных изменений напряжения. Такое поведение присуще любой т.н. индуктивной нагрузке (трансформаторы, электродвигатели, дроссели, электромагниты).

Помимо индуктивной нагрузки существует емкостная (различные электронные устройства с конденсаторами, как накопителями энергии, например, в импульсном блоке питания), в которой ток, наоборот, опережает напряжение за счёт временного накопления энергии конденсаторами и последующей отдачи её назад. И в том и в другом случае в цепи помимо активной возникает реактивная энергия.

Вред реактивной энергии в электроэнергетике очевиден — она никак не используется, но шляется туда-сюда по проводам, дополнительно нагружая их. Кроме того, при таком «шлянии» эта энергия ещё и частично теряется, преобразуясь в активную энергию при нагреве проводов. Однако в радиотехнике реактивная мощность может быть и полезной (например, в колебательных контурах).

Реактивная мощность обозначается буквой Q и измеряется в вольт-амперах реактивных (вар).

Для вычисления доли реактивной мощности применяется формула:

Q = U * I * sin φ, где:
sin φ — коэффициент мощности, показывающий, какую долю полной мощности составляет реактивная мощность.

Для вычисления активной мощности в сетях с реактивной составляющей применяется формула:

P = U * I * cos φ, где:
cos φ — коэффициент мощности, показывающий, какую долю полной мощности составляет активная мощность.

Коэффициенты мощностей разных приборов обычно указываются в паспортах на них.

**Полная мощность**

Полная мощность — эта вся мощность, и активная и неактивная.

Полная мощность обозначается буквой S и измеряется в вольт-амперах (ВА).

В случае линейной (равномерной на протяжении периода) нагрузки полная мощность равна корню квадратному из суммы квадратов активной и реактивной мощности. В этом случае неактивная мощность полностью состоит из реактивной составляющей.

S = √(P² + Q²)

То есть, полная мощность получается не лобовым сложением активной и неактивной частей, а по закону прямоугольного треугольника:

Попробуем определить количество центроидов для столбца bulk_14.

```python
data = data_bulk_full['bulk_14'].dropna()
max_clusters = 20

# Преобразование данных в формат, подходящий для k-средних
X = data.values.reshape(-1, 1)

# Список для сохранения инерции (суммы квадратов расстояний от точек до центроидов) для каждого количества кластеров
inertia = []

# Поиск инерции для разного количества кластеров от 1 до max_clusters
for n_clusters in range(1, max_clusters + 1):
    kmeans = KMeans(n_clusters=n_clusters)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

# Визуализация метода "локтя"
plt.figure(figsize=(15,12))
plt.plot(range(1, max_clusters + 1), inertia, marker='o')
plt.xlabel('Количество кластеров')
plt.ylabel('Значение целевой функции')
plt.title('Метод "локтя" для определения оптимального количества кластеров')
plt.show()
```

На уровне 8 центроидов график более-менее выходит на плато.

```python
n_clusters = 12


# Создание и обучение модели k-средних
kmeans = KMeans(n_clusters=n_clusters)
kmeans.fit(X)

# Получение центроидов (средних значений) каждого кластера
centroids = kmeans.cluster_centers_

# Получение меток кластеров для каждого элемента данных
labels = kmeans.labels_

# Вывод результатов
print("Центроиды (средние значения) кластеров:")
print(centroids)

# Визуализация результатов
plt.figure(figsize=(15,15))
plt.scatter(data.index, data.values, c=labels, cmap='viridis', marker='o', s=10)
plt.scatter(range(n_clusters), centroids, c='red', marker='o', s=200, label='Центроиды')
plt.xlabel('Индекс элемента данных')
plt.ylabel('Значение легирующего элемента')
plt.legend()
plt.show()
```

Центроиды подозрительно округленно кратны то-ли к 25, то ли к 50.

Добавление легирующих элементов в сталь происходит с целью изменения её свойств и характеристик для соответствия определенным требованиям и задачам. Каждый легирующий элемент имеет свои уникальные свойства и влияет на структуру и свойства стали. Ниже приведены некоторые примеры легирующих элементов и их обычные пределы добавления в сталь:

Углерод (С): Обычно присутствует в стали в пределах от 0,02% до 2,1%. Углерод влияет на твердость и прочность стали.

Марганец (Mn): Обычно добавляется в пределах от 0,25% до 2%. Марганец улучшает ударную вязкость и образует твердые растворы в структуре стали.

Кремний (Si): Добавляется в пределах от 0,15% до 0,5%. Кремний улучшает прокатываемость и уменьшает окисление стали.

Сера (S): Присутствует в небольших количествах, обычно не более 0,05%. Сера повышает обработку стали и влияет на её свариваемость.

Фосфор (P): Присутствует в небольших количествах, обычно не более 0,05%. Фосфор улучшает свариваемость и обрабатываемость стали.

Хром (Cr): Обычно добавляется в пределах от 0,5% до 2%. Хром улучшает стойкость к коррозии и повышает твердость.

Никель (Ni): Добавляется в пределах от 0,5% до 5%. Никель улучшает ударную вязкость, стойкость к коррозии и магнитные свойства.

Молибден (Mo): Добавляется в пределах от 0,1% до 0,5%. Молибден улучшает термическую стойкость и стойкость к растрескиванию.

Ванадий (V): Добавляется в пределах от 0,1% до 0,5%. Ванадий увеличивает прочность и твердость.

Ниобий (Nb): Добавляется в пределах от 0,02% до 0,2%. Ниобий улучшает ударную вязкость и термическую стойкость.

Это лишь общие примеры легирующих элементов и их пределов добавления в сталь. Фактические значения могут отличаться в зависимости от конкретного типа стали, её предназначения и требований к свойствам. ## Обучение моделей

### Создание функций и разделение на выборки

Для равности условий при выполнении проекта random_state решено сделать равным 30723.

```python
rs = 30723
```

Сделаем функцию для разделения выборок на обучающую и тестовую 75:25.

```python
def split_data(df, target):
    data_train, data_test = train_test_split(df, test_size=0.25, random_state=rs)
    
    train_features = data_train.drop([target], axis=1)
    train_target = data_train[target]
    
    test_features = data_test.drop([target], axis=1)
    test_target = data_test[target]
    
    return train_features, train_target, test_features, test_target
```

Сделаю функцию для поиска наилучших гиперпараметров через GridSearch.

```python
def grid_search(model, params):
    grid_cv = GridSearchCV(estimator=model, param_grid=params, cv=10, n_jobs=-1, scoring='neg_mean_absolute_error')
    grid_cv.fit(train_features, train_target)
    print('Гиперпараметры:', grid_cv.best_params_)
    print('MAE:', round(-grid_cv.best_score_, 2))
    

    return grid_cv.best_params_
```

```python
columns = ['model_name', 'model', 'MAE', 'dataset']
result_tbl = pd.DataFrame(columns=columns)
result_tbl
```

```python
def test(model, name, dataset):
    
    model.fit(train_features, train_target)
    
    predictions = model.predict(test_features)
    global a
    columns = ['model_name', 'model', 'MAE', 'dataset']
    rows = [[name, model,  round(mean_absolute_error(test_target, predictions), 2), dataset]]
    a = pd.DataFrame(data=rows, columns=columns)
```

```python
df_arc = pd.DataFrame(data_arc)


result_df_arc = pd.DataFrame(columns=df_arc.columns)


for key_value in df_arc['key'].unique():
    # Отбор данных для текущего ключа
    temp_df = df_arc[df_arc['key'] == key_value].copy()
    # Создание пустой строки для текущего ключа
    empty_row = {col: 0 for col in temp_df.columns}
    empty_row['key'] = key_value
    empty_row_df = pd.DataFrame([empty_row])
    # Объединение пустой строки с данными для текущего ключа
    temp_df = pd.concat([empty_row_df, temp_df], ignore_index=True)
    # Добавление результатов в итоговый DataFrame
    result_df_arc = pd.concat([result_df_arc, temp_df], ignore_index=True)
result_df_arc.head(15)
```

Наверное, таким был первоначальный вид таблицы(но это не точно:-))

Для дальнейшей стыковки таблиц нужно добавить дополнительные кдючи внутри таблицы. Но есть одно НО, во время анализа были выявлены случаи(а их не мало), когда количество прогревов было больше чем количество добавляемых легирующих элементови и "холостые" прогревы были в начале партий. Дальнейшие действия будут идти от предположения, что все холостые прогревы только в начале партии.  

```python
melted_df['key2'] = melted_df.groupby('key')['time'].rank(method='dense').astype(int)
melted_df.head(15)
```

```python
melted_df.reset_index(inplace=True)
melted_df
```

```python
melted_df.columns = ['key', 'name_elem_time', 'time', 'name_elem', 'volume', 'key2']
```

```python
melted_df.head(15)
```

Теперь развернем столбцы обратно, но при этом сохранив строки с ключами. Из "базовой таблицы сделаем две, отдельно для временных и отдельно для объемов. Потом объединим их. 

```python
pivoted_df_time = melted_df.pivot(index=['key', 'key2'], columns='name_elem_time', values='time')
pivoted_df_time = pivoted_df_time.sort_values(by=['key', 'key2'], ascending=[True, True])
pivoted_df_time.reset_index(inplace=True)
pivoted_df_time.columns.name = None
time_columns = ['key', 'key2', 'bulk_1_time', 'bulk_2_time', 'bulk_3_time', 'bulk_4_time', 'bulk_5_time',
                   'bulk_6_time', 'bulk_7_time', 'bulk_8_time', 'bulk_9_time', 'bulk_10_time', 'bulk_11_time',
                   'bulk_12_time', 'bulk_13_time', 'bulk_14_time', 'bulk_15_time', 'wire_1_time', 'wire_2_time',
                   'wire_3_time', 'wire_4_time', 'wire_5_time', 'wire_6_time', 'wire_7_time', 'wire_8_time', 'wire_9_time']


pivoted_df_time_new = pivoted_df_time.reindex(columns=time_columns)


pivoted_df_time_new.head(15)
```

Как в предыдущем варианте проекта сделаем порядок добавления легирующих элементов. Перед этим заполним пустующие значения внутри каждой партии предыдущим значением. 

```python


time_columns = ['bulk_1_time', 'bulk_2_time', 'bulk_3_time', 'bulk_4_time', 'bulk_5_time',
                'bulk_6_time', 'bulk_7_time', 'bulk_8_time', 'bulk_9_time', 'bulk_10_time',
                'bulk_11_time', 'bulk_12_time', 'bulk_13_time', 'bulk_14_time', 'bulk_15_time',
                'wire_1_time', 'wire_2_time', 'wire_3_time', 'wire_4_time', 'wire_5_time',
                'wire_6_time', 'wire_7_time', 'wire_8_time', 'wire_9_time']

pivoted_df_time[time_columns] = pivoted_df_time_new.groupby('key')[time_columns].fillna(method='ffill')
pivoted_df_time.head(15)
```

Произведем ранжирование по порядку добавления легирующего элемента

```python


pivoted_df_time[time_columns] = pivoted_df_time[time_columns].apply(lambda row: row.rank(na_option='keep'), axis=1)
pivoted_df_time = pivoted_df_time.fillna(0)
pivoted_df_time
```

Вроде все получилось, теперь перейдем ко второй части базовой таблицы. При этом на каждой итерации внутри партий будут учитываться объемы всех л.э. которые уже добавлялись ранее.

```python
pivoted_df_volume = melted_df.pivot(index=['key', 'key2'], columns='name_elem', values='volume')
pivoted_df_volume = pivoted_df_volume.sort_values(by=['key', 'key2'], ascending=[True, True])
pivoted_df_volume.reset_index(inplace=True)
pivoted_df_volume.columns.name = None
volume_columns = ['key', 'key2', 'bulk_1', 'bulk_2', 'bulk_3', 'bulk_4', 'bulk_5',
                   'bulk_6', 'bulk_7', 'bulk_8', 'bulk_9', 'bulk_10', 'bulk_11',
                   'bulk_12', 'bulk_13', 'bulk_14', 'bulk_15', 'wire_1', 'wire_2',
                   'wire_3', 'wire_4', 'wire_5', 'wire_6', 'wire_7', 'wire_8', 'wire_9']


pivoted_df_volume = pivoted_df_volume.replace(0, pd.NA)
pivoted_df_volume[volume_columns] = pivoted_df_volume.groupby('key')[volume_columns].fillna(method='ffill')

pivoted_df_volume = pivoted_df_volume.reindex(columns=volume_columns)
pivoted_df_volume = pivoted_df_volume.fillna(0)

pivoted_df_volume.head(15)
```

Объединим всё обратно

```python
def test(model, name, dataset):
    
    model.fit(train_features, train_target)
    
    predictions = model.predict(test_features)
    global a
    columns = ['model_name', 'model', 'MAE', 'dataset']
    rows = [[name, model,  round(mean_absolute_error(test_target, predictions), 2), dataset]]
    a = pd.DataFrame(data=rows, columns=columns)
```
